Crie as classes "**Livro**" e "**Biblioteca**". A classe Biblioteca deve conter uma lista de objetos do tipo Livro. Implemente métodos para adicionar livros e listar todos os títulos disponíveis.

```python
class Book:
  def __init__(self, name: str):
    self.name = name

class Library:
  def __init__(self, books: list[Book]):
    self.books = books

  def add_book(self, new_book: Book):
    self.books.append(new_book)

  def list_books(self):
    for book in self.books:
      print(f"- {book.name}")

library: Library = Library([
  Book("Harry Potter"),
  Book("O Senhor dos Anéis"),
  Book("The Hobbit")
])

library.add_book(Book("Game of Thrones"))

print("------------------------------------------------------------------")
print("LIVROS")
print("------------------------------------------------------------------")

library.list_books()

print("------------------------------------------------------------------")
```
