### 02 - Importação (Recursos)

```python
import numpy as np
import matplotlib.pyplot as plt

print(f"Numpy Version: {np.__version__}")
```

### 03 - Carregando o Arquivo CSV

```python
try:
    data = np.genfromtxt("./dataset_clientes_kmeans.csv", delimiter=",", skip_header=1)

    print(data)
except:
    print("Erro ao carregar o arquivo CSV!")

    exit()
```

### 04 - Funções

```python
def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2, axis=1))

def kmeans(X, k=3, max_iter=100):
    idx = np.random.choice(len(X), k, replace=False)

    centroids = X[idx]

    for _ in range(max_iter):
        clusters = [[] for _ in range(k)]

        labels = np.zeros(len(X))

        for i, point in enumerate(X):
            distances = euclidean_distance(point, centroids)

            label = np.argmin(distances)

            clusters[label].append(point)

            labels[i] = label

        old_centroids = centroids.copy()

        for i in range(k):
            if (clusters[i]):
                centroids[i] = np.mean(clusters[i], axis=0)

        if (np.allclose(centroids, old_centroids)):
            break

    return centroids, labels
```

### 05 - Gráfico

```python
centroids, labels = kmeans(data, k=3)

plt.figure(figsize=(10,6))

for cluster_id in np.unique(labels):
    plt.scatter(data[labels == cluster_id, 0], data[labels == cluster_id, 1], alpha=0.7, label=f"Cluster ({int(cluster_id)})")

plt.scatter(centroids[:, 0], centroids[:, 1], s=180, c="black", marker="X", label="Centróides")

plt.title("Segmentação de Clientes (K-Means)")

plt.xlabel("Frequência de Compras (Mês)")
plt.ylabel("Ticket Médio (R$)")

plt.legend()

plt.grid(True)

plt.show()
```
