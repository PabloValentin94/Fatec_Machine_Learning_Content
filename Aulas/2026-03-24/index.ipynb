### 03 - Regressão Linear Múltipla

```python
X = np.array([
    [1, 50, 2],
    [1, 80, 3],
    [1, 100, 4]
])

y = np.array([200, 300, 400])

# Obtendo o valor do coeficiente beta.

beta = np.linalg.inv(X.T @ X) @ X.T @ y

print(beta)
```

```python
# Lendo o arquivo CSV.

data = np.genfromtxt("./kc_house_data.csv", delimiter=",", skip_header=1, usecols=(2, 3, 4, 5))

# Obtendo a coluna de preços.

y = data[:, 0]

# Obtendo as colunas restantes da análise.

X_features = data[:, 1:]

# Quantidade de linhas das features.

m = X_features.shape[0]

# Calculando a matriz X.

X = np.c_[np.ones(m), X_features]

# Matriz transposta de X.

X_T = X.T

# Obtendo o valor do coeficiente beta.

beta = np.linalg.inv(X_T @ X) @ X_T @ y

# Saída (Beta).

print("=============== COEFICIENTES DESCOBERTOS ===============")
print(f"Intercepto (Beta 0): {beta[0]:.2f}")
print(f"Peso para Quartos (Beta 1): {beta[1]:.2f}")
print(f"Peso para Banheiros (Beta 2): {beta[2]:.2f}")
print(f"Peso para Área/SQFT (Beta 3): {beta[3]:.2f}\n")

# Testando.

test = np.array([1, 3, 1, 1180])

predicted_price = test @ beta

# Saída (Teste).

print("=============== PREVISÃO (TESTE) ===============")
print(f"Preço real da casa 01: ${y[0]:.2f}")
print(f"Preço previsto da casa 01: {predicted_price:.2f}")
```
