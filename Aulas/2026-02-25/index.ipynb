```python
import matplotlib.pyplot as plt

years = [2020, 2021, 2022, 2023, 2024]

sales = [100, 120, 150, 130, 160]

plt.plot(years, sales)

plt.title("Vendas ao Longo dos Anos")

plt.xlabel("Anos")
plt.ylabel("Vendas (Milhão)")

plt.grid(True)

plt.show()
```
