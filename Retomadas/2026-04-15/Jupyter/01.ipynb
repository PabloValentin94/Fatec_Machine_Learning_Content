### 02 - Importação (Recursos)

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

print(f"Numpy Version: {np.__version__}")
```

### 03 - Preparação do Dataset

```python
# Obtenção e tratamento de dados.

dataset = np.loadtxt(
  fname="../Material/knn_dataset_exercicio_1.txt",
  dtype=float,
  delimiter="|",
  skiprows=2,
  usecols=(1,2,3),
)

# Segmentação de dados.

X_train = dataset[:, :2].astype(float)

y_train = dataset[:, 2].astype(int)

print(f"Amostra das Características:\n{X_train}")

print(f"\nAmostra dos Rótulos:\n{y_train}")
```

### 04 - Visualização de Dados de Treinamento

```python
# Estilo de fundo do gráfico.

plt.style.use("dark_background")

# Definição da estrutura do gráfico.

fig, ax = plt.subplots(figsize=(10,7))

# Classificações disponíveis (Rótulos).

ax.scatter(X_train[y_train == 1, 0], X_train[y_train == 1, 1], c="blue", s=100, edgecolor="k", alpha=0.7, label="Baixo Consumo")
ax.scatter(X_train[y_train == 2, 0], X_train[y_train == 2, 1], c="red", s=100, edgecolor="k", alpha=0.7, label="Médio Consumo")
ax.scatter(X_train[y_train == 3, 0], X_train[y_train == 3, 1], c="purple", s=100, edgecolor="k", alpha=0.7, label="Alto Consumo")

# Textos informativos do gráfico.

ax.set_title("Padrão de Consumo dos Clientes:", fontsize=16)

ax.set_xlabel("Renda Mensal", fontsize=12)
ax.set_ylabel("Gasto Mensal", fontsize=12)

# Habilitando as legendas do gráfico.

ax.legend()

# Exibindo o gráfico.

plt.show()
```

### 05 - Funções do KNN

```python
def calculate_euclidean_distance(point_1, point_2):
    return np.sqrt(np.sum((point_1 - point_2) ** 2))

def find_neighboors(X_train, y_train, test_point, k):
    distances = []

    for i, train_point in enumerate(X_train):
        distance = calculate_euclidean_distance(train_point, test_point)

        distances.append((distance, y_train[i]))

    distances.sort(key=(lambda x: x[0]))

    neighboors = [distance[1] for distance in distances[:k]]

    return neighboors

def preview_classification(neighboors):
    votes_countage = Counter(neighboors)

    preview = votes_countage.most_common(1)[0][0]

    return preview
```

### 06 - Aplicação de Teste

```python
new_client = np.array([6000.0, 3000.0])

k = 5

nearby_neighboors = find_neighboors(X_train, y_train, new_client, k)

final_preview = preview_classification(nearby_neighboors)

result_label = ""

if (final_preview == 1):
  result_label = "Baixo Consumo"
elif (final_preview == 2):
  result_label = "Médio Consumo"
else:
  result_label = "Alto Consumo"

print(f"Características:\n{new_client}")

print(f"\nValor de K:\n{k}")

print(f"\nClasses dos Vizinhos Próximos:\n{nearby_neighboors}")

print(f"\nClasse do Novo Cliente:\n{result_label} ({final_preview})")
```

### 07 - Gráfico de Teste

```python
fig, ax = plt.subplots(figsize=(12, 8))

ax.scatter(X_train[y_train == 1, 0], X_train[y_train == 1, 1], c="blue", s=100, edgecolor="k", alpha=0.7, label="Baixo Consumo (Classe 1)")
ax.scatter(X_train[y_train == 2, 0], X_train[y_train == 2, 1], c="red", s=100, edgecolor="k", alpha=0.7, label="Médio Consumo (Classe 2)")
ax.scatter(X_train[y_train == 3, 0], X_train[y_train == 3, 1], c="purple", s=100, edgecolor="k", alpha=0.7, label="Alto Consumo (Classe 3)")

ax.scatter(new_client[0], new_client[1], c="orange", s=200, edgecolor="k", marker="*", label="Novo Cliente")

neighboors_indexes = np.argsort([calculate_euclidean_distance(index, new_client) for index in X_train])[:k]

neighboors_points = X_train[neighboors_indexes]

maximum_distance = calculate_euclidean_distance(new_client, neighboors_points[-1])

circle = plt.Circle(new_client, maximum_distance, color="yellow", fill=False, linestyle="--", linewidth=2, label=f"Vizinhança (K = {k})")

ax.add_artist(circle)

ax.set_title("Resultado da Classificação (KNN)", fontsize=16)

ax.set_xlabel("Renda Mensal", fontsize=12)
ax.set_ylabel("Gasto Mensal", fontsize=12)

ax.legend(loc="upper left")

ax.grid(True)

plt.show()
```
