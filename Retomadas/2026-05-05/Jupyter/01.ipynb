### 03 - Preparação do Dataset

```python
def load_txt_data(file_path):
    X, y = [], []

    dataset = np.loadtxt(
        fname=file_path,
        dtype=int,
        delimiter="|",
        skiprows=2,
        usecols=(1,2,3,4)
    )

    X = dataset[:, :3]

    y = dataset[:, 3]

    return X, y

X, y = load_txt_data("../Material/arvore_decisao_dataset_exercicio_1.txt")

print(f"Amostras:\n{X}")
print(f"\nRótulos:\n{y}")
```
