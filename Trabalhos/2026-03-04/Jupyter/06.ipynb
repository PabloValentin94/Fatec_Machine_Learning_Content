# Exercício 06

No contexto de IoT, crie uma classe "**Sensor**" que armazene "**ID**" e "**tipo**". Crie um script que permita ao usuário cadastrar vários sensores em um dicionário, onde a chave é o ID e o valor é o objeto Sensor.

```python
class Sensor:
  def __init__(self, id: int, type: str):
    self.id = id
    self.type = type

sensors: dict[int, Sensor] = {}

def add_sensor(sensor: Sensor):
  sensors[sensor.id] = sensor

def list_sensors():
  for key in sensors:
    print(f"{key} - {sensors[key].type}")

add_sensor(Sensor(1, "Proximidade"))
add_sensor(Sensor(2, "Movimento"))
add_sensor(Sensor(3, "Fotossensível"))

print("------------------------------------------------------------------")
print("SENSORES")
print("------------------------------------------------------------------")

list_sensors()

print("------------------------------------------------------------------")
```
