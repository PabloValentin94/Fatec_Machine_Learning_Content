### 07 - Seleção e Filtragem de DataFrames

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

titanic_dataframe = pandas.read_csv(url, delimiter=",")

ages = titanic_dataframe["Age"] # Retorna uma nova Series.

display(ages)

filtered_dataframe = titanic_dataframe[["Name", "Age", "Survived"]] # Retorna um novo DataFrame.

display(filtered_dataframe)

elderly_passengers = titanic_dataframe[titanic_dataframe["Age"] > 60] # Filtragem condicional.

display(elderly_passengers)

multiple_filter = (titanic_dataframe["Sex"] == "female") & (titanic_dataframe["Survived"] == 1) & (titanic_dataframe["Pclass"] == 1)

rich_women_passengers = titanic_dataframe[multiple_filter] # Filtragem múltipla.

display(rich_women_passengers)
```
