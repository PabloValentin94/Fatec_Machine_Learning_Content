Crie uma classe base "**Pagamento**" com um método "**processar()**". Crie as subclasses "**PagamentoCartao**" e "**PagamentoBoleto**". Cada uma deve implementar o método processar() exibindo uma mensagem específica de como o valor será debitado.

```python
class Payment:
  def __init__(self):
    pass

  def process(self):
    pass

class CardPayment(Payment):
  def __init__(self):
    pass

  def process(self):
    print("Efetuando pagamento no cartão.")

class BankSlipPayment(Payment):
  def __init__(self):
    pass

  def process(self):
    print("Efetuando pagamento no boleto.")

card_payment: CardPayment = CardPayment()

bank_slip_payment: BankSlipPayment = BankSlipPayment()

print("------------------------------------------------------------------")
print("PAGAMENTO - CARTÃO")
print("------------------------------------------------------------------")

card_payment.process()

print("------------------------------------------------------------------")
print("PAGAMENTO - BOLETO")
print("------------------------------------------------------------------")

bank_slip_payment.process()

print("------------------------------------------------------------------")
```
