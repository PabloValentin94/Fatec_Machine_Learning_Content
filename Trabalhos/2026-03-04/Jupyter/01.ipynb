# Exercício 01

Crie uma classe "**Aluno**" que armazene um nome e uma lista de notas. Implemente um método para calcular a média e outro para verificar se o aluno está aprovado, ou seja, sua média é maior ou igual a 6.

```python
class Student:
  def __init__(self, name: str, notes: list[float]):
    self.name = name
    self.notes = notes

  def verify_student_situation(self, media: float):
    if media >= 6:
      return "Aprovado"
    else:
      return "Reprovado"

  def calculate_media(self):
    media: float = (sum(self.notes)) / len(self.notes)

    print(f"A média do aluno(a) {self.name} é: {media:.2f}")

    print("------------------------------------------------------------------")

    print(f"Situação do aluno(a): {self.verify_student_situation(media)}.")

student: Student = Student("Pablo Valentin", [10, 6, 8])

print("------------------------------------------------------------------")

student.calculate_media()

print("------------------------------------------------------------------")
```
