```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

print(f"Numpy Version: {np.__version__}")
```

```python
dataset = np.loadtxt("../Data/03.txt", dtype=str, delimiter=",", skiprows=1, usecols=(1,2,3))

X_train = dataset[:, :2].astype(int)

y_train = dataset[:, 2].astype(str)

print(f"Amostra das Características:\n{X_train[:5]}")

print(f"\nAmostra dos Rótulos:\n{y_train[:5]}")
```

```python
plt.style.use("seaborn-v0_8-whitegrid")

fig, ax = plt.subplots(figsize=(10, 7))

ax.scatter(X_train[y_train == "A", 0], X_train[y_train == "A", 1], c="blue", s=100, edgecolor="k", alpha=0.7, label="Frequente (A)")

ax.scatter(X_train[y_train == "B", 0], X_train[y_train == "B", 1], c="red", s=100, edgecolor="k", alpha=0.7, label="Ocasional (B)")

ax.set_title("Distribuição de Frequência dos Alunos:", fontsize=16)

ax.set_xlabel("Idade", fontsize=12)

ax.set_ylabel("Gasto Mensal em Suplementos", fontsize=12)

ax.legend()

plt.show()
```

```python
def calculate_euclidean_distance(point_1, point_2):
    return np.sqrt(np.sum((point_1 - point_2) ** 2))

def find_neighboors(X_train, y_train, test_point, k):
    distances = []

    for i, train_point in enumerate(X_train):
        distance = calculate_euclidean_distance(train_point, test_point)

        distances.append((distance, y_train[i]))

    distances.sort(key=(lambda x: x[0]))

    neighboors = [distance[1] for distance in distances[:k]]

    return neighboors

def preview_classification(neighboors):
    votes_countage = Counter(neighboors)

    preview = votes_countage.most_common(1)[0][0]

    return preview
```

```python
new_student = np.array([33, 100.0])

k = 5

nearby_neighboors = find_neighboors(X_train, y_train, new_student, k)

final_preview = preview_classification(nearby_neighboors)

result_label = "Frequente" if (final_preview == "A") else "Ocasional"

print(f"Características:\n{new_student}")

print(f"\nValor de K:\n{k}")

print(f"\nClasses dos Vizinhos Próximos:\n{nearby_neighboors}")

print(f"\nClasse do Novo Aluno:\n{result_label} ({final_preview})")
```

```python
fig, ax = plt.subplots(figsize=(12, 8))

ax.scatter(X_train[y_train == "A", 0], X_train[y_train == "A", 1], c="blue", s=100, edgecolor="k", alpha=0.7, label="Frequente (A)")
ax.scatter(X_train[y_train == "B", 0], X_train[y_train == "B", 1], c="red", s=100, edgecolor="k", alpha=0.7, label="Ocasional (B)")

ax.scatter(new_student[0], new_student[1], c="green", s=200, edgecolor="k", marker="*", label="Novo Aluno")

neighboors_indexes = np.argsort([calculate_euclidean_distance(index, new_student) for index in X_train])[:k]

neighboors_points = X_train[neighboors_indexes]

maximum_distance = calculate_euclidean_distance(new_student, neighboors_points[-1])

circle = plt.Circle(new_student, maximum_distance, color="gray", fill=False, linestyle="--", linewidth=2, label=f"Vizinhança (K = {k})")

ax.add_artist(circle)

ax.set_title("Resultado da Classificação (KNN)", fontsize=16)

ax.set_xlabel("Idade", fontsize=12)
ax.set_ylabel("Gasto Mensal em Suplementos", fontsize=12)

ax.legend(loc="upper left")

ax.grid(True)

plt.show()
```

### **Resposta**:

Escolhemos esse valor (**K = 5**) pois o cinco é um bom número para evitar underfitting (Subajuste), mantendo os dados suavizados, e overfiting (Sobreajuste). Outra razão é que devido ao número ser impar, a chance de empate é quase zero.
