Crie uma classe "**Calculadora**" que contenha apenas "**métodos estáticos**" para operações como "**soma**", "**subtração**" e "**raiz quadrada**". O objetivo é usar a classe como um "namespace" de funções utilitárias sem precisar instanciá-la.

```python
import math

class Calculator:
  def __init__(self):
    pass

  @staticmethod
  def addition(first_number, second_number):
    print(f"Soma: {(first_number + second_number):.2f}")
  
  @staticmethod
  def subtraction(first_number, second_number):
    print(f"\nSubtração: {(first_number - second_number):.2f}")
  
  @staticmethod
  def multiplication(first_number, second_number):
    print(f"\nMultiplicação: {(first_number * second_number):.2f}")
  
  @staticmethod
  def division(first_number, second_number):
    print(f"\nDivisão: {(first_number / second_number):.2f}")

  @staticmethod
  def power(base, exponent):
    print(f"\nPotência: {math.pow(base, exponent):.2f}")

  @staticmethod
  def square_root(number):
    print(f"\nRaíz Quadrada: {math.sqrt(number):.2f}")

print("------------------------------------------------------------------")
print("CALCULADORA")
print("------------------------------------------------------------------")
  
Calculator.addition(5, 5)

Calculator.subtraction(5, 5)

Calculator.multiplication(5, 5)

Calculator.division(5, 5)

Calculator.power(5, 3)

Calculator.square_root(25)

print("------------------------------------------------------------------")
```
