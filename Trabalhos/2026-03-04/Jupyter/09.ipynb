# Exercício 09

Crie uma classe "**Funcionario**" que possua um método "**registrar_entrada()**". Utilize o módulo "**datetime**" para capturar o horário exato em que o método foi chamado e armazene isso em uma lista de logs dentro do objeto.

```python
import time
from datetime import datetime

class Employee:
  def __init__(self, name: str):
    self.name = name
    self.__access_logs: list = []

  def register_entry(self):
    self.__access_logs.append({ "horary": datetime.now() })

  def list_access_logs(self):
    for access_log in self.__access_logs:
      print(f"- Horário de acesso: {access_log['horary']}")

employee: Employee = Employee("Pablo Valentin")

for i in range(0, 5):
  employee.register_entry()
  time.sleep(1) # Espera intencional de um segundo.

print("------------------------------------------------------------------")
print(f"LOGS DE ACESSO DO FUNCIONÁRIO ({employee.name})")
print("------------------------------------------------------------------")

employee.list_access_logs()

print("------------------------------------------------------------------")
```
