# Exercício 08

Crie uma classe "**Camera**" e uma classe "**Telefone**". Crie uma terceira classe "**Smartphone**" que herde de ambas, demonstrando como o objeto final pode acessar métodos das duas classes originais.

```python
class Camera:
  def __init__(self):
    pass

  def take_picture(self):
    print("Uma foto foi tirada.")

class Telephone:
  def __init__(self):
    pass

  def make_call(self):
    print("Uma ligação telefônica foi feita.")

class Smartphone(Telephone, Camera):
  def __init__(self):
    pass

smartphone: Smartphone = Smartphone()

print("------------------------------------------------------------------")
print("FUNÇÕES DE UM SMARTPHONE")
print("------------------------------------------------------------------")

smartphone.take_picture()

print("------------------------------------------------------------------")

smartphone.make_call()

print("------------------------------------------------------------------")
```
