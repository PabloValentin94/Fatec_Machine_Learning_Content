# Exercício 02

Crie uma classe "**Conversor**" que possua um atributo privado "**taxa_dolar**". Implemente métodos para atualizar a taxa e para converter um valor de "Real" para "Dólar". Adicione uma validação para que não sejam aceitos valores negativos.

```python
class Converter:
  def __init__(self):
    self.__dollar_rate = 1 # Atributo privado.

  def update_dollar_rate(self, new_dollar_rate: float):
    if (new_dollar_rate > 0):
      self.__dollar_rate = new_dollar_rate

  def calculate_real_convertion(self, dollar_quantity: float):
    print(f"U${dollar_quantity:.2f} <--> R${(dollar_quantity * self.__dollar_rate):.2f}")

  def calculate_dollar_convertion(self, real_quantity: float):
    print(f"R${real_quantity:.2f} <--> U${(real_quantity / self.__dollar_rate):.2f}")

converter: Converter = Converter()

converter.update_dollar_rate(6.29)

print("------------------------------------------------------------------")
print("CONVERSÃO PARA REAIS")
print("------------------------------------------------------------------")

converter.calculate_real_convertion(5.50)

print("------------------------------------------------------------------")
print("CONVERSÃO PARA DÓLARES")
print("------------------------------------------------------------------")

converter.calculate_dollar_convertion(31.45)

print("------------------------------------------------------------------")

# print(converter.__dollar_rate) # Isso gera um erro, pois o atributo é privado.
```
