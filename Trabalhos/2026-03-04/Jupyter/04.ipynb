# Exercício 04

Crie uma classe base "**Veiculo**" com atributos como "**marca**" e "**modelo**". Crie subclasses "**Carro**" (Contém a quantidade de portas.) e "**Moto**" (Contém a quantidade de cilindradas.). Ambas devem herdar as características da classe pai.

```python
class Vehicle:
  def __init__(self, brand: str, model: str):
    self.brand = brand
    self.model = model

  def start_vehicle(self):
    print("Identificando veículo...\n")

class Car(Vehicle):
  def __init__(self, brand: str, model: str, doors_quantity: int):
    super().__init__(brand, model)
    self.doors_quantity = doors_quantity

  def start_vehicle(self):
    super().start_vehicle()

    print(f"Ligando carro ({self.brand} | {self.model} | {self.doors_quantity} Portas).")

class Motorcycle(Vehicle):
  def __init__(self, brand: str, model: str, engine_displacements: float):
    super().__init__(brand, model)
    self.engine_displacements = engine_displacements

  def start_vehicle(self):
    super().start_vehicle()

    print(f"Ligando moto ({self.brand} | {self.model} | {self.engine_displacements}cc).")

car: Car = Car("Volkswagen", "Golf", 4)

motorcycle: Motorcycle = Motorcycle("Honda", "CB500", 471)

print("------------------------------------------------------------------")
print("CARRO")
print("------------------------------------------------------------------")

car.start_vehicle()

print("------------------------------------------------------------------")
print("MOTO")
print("------------------------------------------------------------------")

motorcycle.start_vehicle()

print("------------------------------------------------------------------")
```
