### 02 - Importação (Recursos)

```python
import numpy as np

print(f"Numpy Version: {np.__version__}")
```

### 03 - Preparação do CSV

```python
try:
    data = np.genfromtxt("./kc_house_data.csv", delimiter=",", skip_header=1, usecols=(2,3,4,5,17,18))

    print(data)
except:
    print("Erro ao carregar o CSV!")

    exit()
```

### 04 - Preparação dos Dados

```python
# Colunas relevantes para o cálculo.

y = data[:, 0] # Price

X_features = data[:, 1:] # Bedrooms, Bathrooms e SQFT Living.

# Quantidade de valores das colunas.

sample_length = X_features.shape[0]

# Cálculo da matriz X.

X = np.c_[np.ones(sample_length), X_features]

# Obtendo a matriz transposta de X.

X_T = X.T
```

### 05 - Coeficientes do Cálculo da Regressão Linear Múltipla

```python
# Obtendo os coeficientes beta.

beta = np.linalg.inv(X_T @ X) @ X_T @ y

# Saída.

print(f"Coeficientes Beta: {beta}")

# Coeficientes obtidos.

print(f"\nIntercepto (Beta 0): {beta[0]:.2f}")
print(f"Peso para 'Bedrooms' (Beta 1): {beta[1]:.2f}") # Quartos.
print(f"Peso para 'Bathrooms' (Beta 2): {beta[2]:.2f}") # Banheiros.
print(f"Peso para 'SQFT Living' (Beta 3): {beta[3]:.2f}") # Área.
print(f"Peso para 'Lat' (Beta 4): {beta[4]:.2f}") # Latitude.
print(f"Peso para 'Long' (Beta 5): {beta[5]:.2f}") # Longitude.
```

### 06 - Teste de Previsão (01)

```python
first_test_sample = np.array([1, 3, 1, 1180, 47.5112, -122.257])

first_predicted_price = first_test_sample @ beta

print(f"Preço Real (Imóvel 01): {y[0]:.2f}")

print(f"\nPreço Estimado (Imóvel 01): {first_predicted_price:.2f}")

first_residual_error = y[0] - first_predicted_price

print(f"\nErro Residual: {first_residual_error:.2f}")
```

### 07 - Teste de Previsão (02)

```python
second_test_sample = np.array([1, 3, 1.75, 1370, 47.6127, -122.045])

second_predicted_price = second_test_sample @ beta

print(f"Preço Real (Imóvel 14): {y[13]:.2f}")

print(f"\nPreço Estimado (Imóvel 14): {second_predicted_price:.2f}")

second_residual_error = y[13] - second_predicted_price

print(f"\nErro Residual: {second_residual_error:.2f}")
```
