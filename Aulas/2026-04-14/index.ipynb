### 02 - Importação (Recursos)

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

print(f"Numpy Version: {np.__version__}")
```

### 03 - Preparação do Dataset

```python
dataset = np.loadtxt("tumores.csv", delimiter=",", skiprows=2)

X_train = dataset[:, :2]

y_train = dataset[:, 2].astype(int)

print(f"Amostra das Características:\n{X_train[:5]}")

print(f"\nAmostra dos Rótulos:\n{y_train[:5]}")
```

### 04 - Visualização de Dados de Treinamento

```python
plt.style.use("seaborn-v0_8-whitegrid")

fig, ax = plt.subplots(figsize=(10, 7))

ax.scatter(X_train[y_train == 0, 0], X_train[y_train == 0, 1], c="blue", s=100, edgecolor="k", alpha=0.7, label="Benigno (0)")

ax.scatter(X_train[y_train == 1, 0], X_train[y_train == 1, 1], c="red", s=100, edgecolor="k", alpha=0.7, label="Maligno (1)")

ax.set_title("Distribuição dos Tumores:", fontsize=16)

ax.set_xlabel("Tamanho do Tumor (CM)", fontsize=12)

ax.set_ylabel("Pontuação de Textura", fontsize=12)

ax.legend()

plt.show()
```

### 05 - Funções do KNN

```python
def calculate_euclidean_distance(point_1, point_2):
    return np.sqrt(np.sum((point_1 - point_2) ** 2))

def find_neighboors(X_train, y_train, test_point, k):
    distances = []

    for i, train_point in enumerate(X_train):
        distance = calculate_euclidean_distance(train_point, test_point)

        distances.append((distance, y_train[i]))

    distances.sort(key=(lambda x: x[0]))

    neighboors = [distance[1] for distance in distances[:k]]

    return neighboors

def preview_classification(neighboors):
    votes_countage = Counter(neighboors)

    preview = votes_countage.most_common(1)[0][0]

    return preview
```

### 06 - Aplicação de Teste

```python
new_tumor = np.array([5.5, 6.0])

k = 5

nearby_neighboors = find_neighboors(X_train, y_train, new_tumor, k)

final_preview = preview_classification(nearby_neighboors)

result_label = "Benigno" if (final_preview == 0) else "Maligno"

print(f"Características:\n{new_tumor}")

print(f"\nValor de K:\n{k}")

print(f"\nClasses dos Vizinhos Próximos:\n{nearby_neighboors}")

print(f"\nClasse do Novo Tumor:\n{result_label} ({final_preview})")
```

### 07 - Gráfico do Teste

```python
fig, ax = plt.subplots(figsize=(12, 8))

ax.scatter(X_train[y_train == 0, 0], X_train[y_train == 0, 1], c="blue", s=100, edgecolor="k", alpha=0.7, label="Benigno (Classe 0)")
ax.scatter(X_train[y_train == 1, 0], X_train[y_train == 1, 1], c="red", s=100, edgecolor="k", alpha=0.7, label="Maligno (Classe 1)")

ax.scatter(new_tumor[0], new_tumor[1], c="green", s=200, edgecolor="k", marker="*", label="Novo Tumor")

neighboors_indexes = np.argsort([calculate_euclidean_distance(index, new_tumor) for index in X_train])[:k]

neighboors_points = X_train[neighboors_indexes]

maximum_distance = calculate_euclidean_distance(new_tumor, neighboors_points[-1])

circle = plt.Circle(new_tumor, maximum_distance, color="gray", fill=False, linestyle="--", linewidth=2, label=f"Vizinhança (K = {k})")

ax.add_artist(circle)

ax.set_title("Resultado da Classificação (KNN)", fontsize=16)

ax.set_xlabel("Tamanho do Tumor (CM)", fontsize=12)

ax.set_ylabel("Pontuação de Textura", fontsize=12)

ax.legend(loc="upper left")

ax.grid(True)

plt.show()
```
