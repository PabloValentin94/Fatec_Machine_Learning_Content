# Exercício 07

Crie uma classe "**Usuario**" com "**login**" e "**senha**". Implemente um método que valide se a senha possui pelo menos 8 caracteres e um caractere especial. Se não atender, o sistema deve solicitar uma nova senha.

```python
import re

class User:
  def __init__(self, login: str, password: str):
    self.login = login
    self.password = password

def verify_password_pattern(user: User):
  if (len(user.password) >= 8) and (re.search(r"[^a-zA-Z0-9\s]", user.password)):
    print(f"A senha do usuário(a) {user.login} é válida.")
  else:
    print(f"A senha do usuário(a) {user.login} é inválida! Defina outra.")

user: User = User("Valentin", "@12345678")

print("------------------------------------------------------------------")
print("VERIFICAÇÃO DE PADRÃO DE SENHA")
print("------------------------------------------------------------------")

verify_password_pattern(user)

print("------------------------------------------------------------------")
```
