### 02 - Importação (Recursos)

```python
import numpy as np
import matplotlib.pyplot as plt

print(f"Numpy Version: {np.__version__}")
```

### 03 - Carregando o Arquivo CSV

```python
try:
    data = np.loadtxt("./dataset.csv", delimiter=",", skiprows=1)

    print(data)
except:
    print("Erro ao carregar o arquivo CSV!")

    exit()
```

### 04 - Separando os Dados

```python
X = data[:, 0] # Constante.

print(X)

y = data[:, 1] # Variável.

print(f"\n{y}")
```

### 05 - Calculando as Colunas da Tabela de Regressão Linear

```python
n = len(X)

print(f"\nNúmero de Amostras: {n}")

x_sum = np.sum(X)

print(f"Soma dos Itens de X: {x_sum}")

y_sum = np.sum(y)

print(f"\nSoma dos Itens de Y: {y_sum}")

square_x_sum = np.sum(X ** 2)

print(f"\nSoma dos Quadrados de X: {square_x_sum}")

square_y_sum = np.sum(y ** 2)

print(f"\nSoma dos Quadrados de Y: {square_y_sum}")

x_y_sum = np.sum(X * y)

print(f"\nSoma dos Itens de X Vezes os Itens de Y: {x_y_sum}")
```

### 06 - Calculando os Coeficientes "A" e "B"

```python
b = (n * (x_y_sum) - (x_sum * y_sum)) / (n * (square_x_sum) - (x_sum ** 2))

print(f"Valor do Coeficiente B: {b}")

a = (y_sum - (b * x_sum)) / n

print(f"\nValor do Coeficiente A: {a}")
```
