```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

print(f"Numpy Version: {np.__version__}")
```

```python
# Lendo o arquivo CSV.

data = np.genfromtxt("../Data/02.txt", delimiter=",", skip_header=1, usecols=(1,2,3,4))

# Obtendo a coluna de preços.

y = data[:, 3]

# Obtendo as colunas restantes da análise.

X_features = data[:, :3]

# Quantidade de linhas das features.

m = X_features.shape[0]

# Calculando a matriz X.

X = np.c_[np.ones(m), X_features]

# Matriz transposta de X.

X_T = X.T

# Obtendo o valor do coeficiente beta.

beta = np.linalg.inv(X_T @ X) @ X_T @ y

# Saída (Beta).

print("=============== COEFICIENTES DESCOBERTOS ===============")
print(f"Intercepto (Beta 0): {beta[0]:.2f}")
print(f"Peso para Quantidade de DEVs (Beta 1): {beta[1]:.2f}")
print(f"Peso para Duração do Projeto em Meses (Beta 2): {beta[2]:.2f}")
print(f"Peso para Complexidade Tecnológica (Beta 3): {beta[3]:.2f}")

# Testando.

test = np.array([0,5,8,7])

predicted_price = test @ beta

# Saída (Teste).

print("\n=============== PREVISÃO (TESTE) ===============")
print(f"Preço Real Calculado (Milhar): R${y[0]:.2f}")
print(f"Preço Justo Estimado (Milhar): R${predicted_price:.2f}")
```
